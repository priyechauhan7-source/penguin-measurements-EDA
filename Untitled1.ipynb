```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Loading the dataset directly from seaborn to avoid FileNotFoundError
df = sns.load_dataset("penguins")

print(df.head())

print("First 5 rows:")
display(df.head())

print("Last 5 rows:")
display(df.tail())

print("Dataset Shape:")
print(df.shape)

print("Dataset Info:")
df.info()

print("Statistical Description:")
display(df.describe())

print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("Species Count:")
print(df["species"].value_counts())

print("\nIsland Count:")
print(df["island"].value_counts())

print("\nSpecies and Island:")
print(df.groupby(["species", "island"]).size())

result = df.groupby("species")[["body_mass_g", "flipper_length_mm"]].mean()
display(result)

average_mass = df.groupby("species")["body_mass_g"].mean()
display(average_mass)

print("Species with highest average body mass:")
print(average_mass.idxmax())

bill_length = df.groupby("species")["bill_length_mm"].mean()
display(bill_length)

columns = ["sex", "bill_length_mm", "bill_depth_mm",
         "flipper_length_mm", "body_mass_g"]

print(df[columns].isnull().sum())

top10 = df.sort_values("body_mass_g", ascending=False).head(10)
display(top10)

df["species"].value_counts().plot(kind="bar")
plt.title("Penguin Count by Species")
plt.xlabel("Species")
plt.ylabel("Count")
plt.tight_layout()
plt.show()

plt.hist(df["body_mass_g"].dropna(), bins=15)
plt.title("Distribution of Penguin Body Mass")
plt.xlabel("Body Mass (grams)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```
